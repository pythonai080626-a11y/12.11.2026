```python
np.random.seed(42)
np1 = np.random.randint(0, 101, (6, 5))
print(np1)
print()

# max(np.arange(1, 10 + 1))  # for 1 dimension

print('max', np1.max())
print('min', np1.min())
print('argmax', np1.argmax())
print('argmin', np1.argmin())
print('shape', np1.shape)

# find row, col of the max use %?  --> 2, 1
# find row, col of the min use %?  --> 3, 1

# step 1- detect how many columns : here, 5
# step 2- 11  % 5 [shape-col] = 1 col
# step 3- 11 // 5 [shape-col] = 2 row

def get_row_col(index, matrix):
    return index // matrix.shape[1], index % matrix.shape[1]

print('max index 11 location', get_row_col(np1.argmax(), np1))
print('max index 16 location', get_row_col(np1.argmin(), np1))
```
